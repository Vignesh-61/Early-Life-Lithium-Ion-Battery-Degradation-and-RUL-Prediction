# 05 — Systematic Memory Window Comparison (W = 3, 5, 10, 20)
Analyzes how historical memory window length impacts degradation prediction accuracy under Leave-One-Battery-Out validation.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt

summary_path = os.path.abspath('../results/phase1/tables/lobo_memory_window_summary.csv')
folds_path = os.path.abspath('../results/phase1/tables/lobo_fold_details.csv')

df_summary = pd.read_csv(summary_path)
df_folds = pd.read_csv(folds_path)
df_summary
```

### Performance Trend Across Window Sizes


```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

# MAE
axes[0].plot(df_summary['window_size'], df_summary['mean_dnn_mae'], marker='o', color='blue', linewidth=2)
axes[0].set_title("Mean Absolute Error (MAE)")
axes[0].set_xlabel("Memory Window (Cycles)")
axes[0].set_ylabel("MAE (Ah)")
axes[0].grid(True, linestyle='--', alpha=0.6)

# RMSE
axes[1].plot(df_summary['window_size'], df_summary['mean_dnn_rmse'], marker='s', color='green', linewidth=2)
axes[1].set_title("Root Mean Squared Error (RMSE)")
axes[1].set_xlabel("Memory Window (Cycles)")
axes[1].set_ylabel("RMSE (Ah)")
axes[1].grid(True, linestyle='--', alpha=0.6)

# MAPE
axes[2].plot(df_summary['window_size'], df_summary['mean_dnn_mape'], marker='^', color='purple', linewidth=2)
axes[2].set_title("Mean Absolute Percentage Error (MAPE)")
axes[2].set_xlabel("Memory Window (Cycles)")
axes[2].set_ylabel("MAPE (%)")
axes[2].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

### Core Research Finding
Notice that **Mean Absolute Error (MAE)** and **MAPE** drop monotonically as the memory window expands from $W=3$ (8.03% MAPE) to $W=20$ (5.20% MAPE).
On well-matched cells such as **B0005**, Memory-20 achieves an exceptional **1.39% MAPE** and $R^2 = 0.976$!
