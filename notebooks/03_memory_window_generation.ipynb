# 03 — Adaptive Memory-Window Generation
This notebook demonstrates how temporal history is converted into fixed-dimensional vectors for feed-forward DNN ingestion without recurrent loops.


```python
import sys
import os
import pandas as pd

sys.path.insert(0, os.path.abspath('../src'))
from memory_windows import create_memory_window_dataset
from preprocessing import compute_soh, compute_rul

df = pd.read_csv(os.path.abspath('../data/processed/cycle_features.csv'))
df = compute_soh(df)
df = compute_rul(df)

for w in [3, 5, 10, 20]:
    ds = create_memory_window_dataset(df, window_size=w)
    print(f"Memory-{w:2d} -> X shape: {ds['X'].shape} | y shape: {ds['y'].shape}")
```
