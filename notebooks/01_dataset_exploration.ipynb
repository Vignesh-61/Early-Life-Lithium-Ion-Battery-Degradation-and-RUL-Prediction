# 01 — NASA Battery Dataset Exploration
**Project:** Early Lithium-Ion Battery Degradation and Remaining Useful Life Prediction Using an Adaptive Memory-Window Deep Neural Network

This notebook explores the raw NASA battery dataset, loads MATLAB `.mat` structures, inspects operational cycles, and plots capacity degradation curves.


```python
import sys
import os
import matplotlib.pyplot as plt

# Add src to path
sys.path.insert(0, os.path.abspath('../src'))
from data_loader import load_all_batteries, get_discharge_capacities

data_dir = os.path.abspath('../data/raw')
batteries = load_all_batteries(data_dir)
```

### Plotting Capacity Degradation Across All 4 Batteries


```python
plt.figure(figsize=(10, 6))
colors = {'B0005': 'blue', 'B0006': 'red', 'B0007': 'green', 'B0018': 'purple'}

for bat_id, bat_data in sorted(batteries.items()):
    cycles, caps = get_discharge_capacities(bat_data)
    plt.plot(cycles, caps, label=f"{bat_id} ({len(caps)} cycles)", color=colors.get(bat_id, 'black'), marker='o', markersize=2.5, alpha=0.85)

plt.axhline(y=1.4, color='gray', linestyle='--', label='EOL Threshold (1.4 Ah)')
plt.title("NASA Li-ion Battery Discharge Capacity Degradation Trajectories", fontsize=13, pad=12)
plt.xlabel("Discharge Cycle Number", fontsize=11)
plt.ylabel("Discharge Capacity (Ah)", fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(framealpha=0.9)
plt.tight_layout()
plt.show()
```

### Observations
- **B0005, B0006, B0007** have 168 discharge cycles, while **B0018** has 132 cycles.
- Notice the **capacity regeneration phenomenon**: periodic upward bumps in capacity due to cell resting and relaxation between discharge cycles.
- B0006 exhibits the fastest degradation rate, dropping below 1.20 Ah.
